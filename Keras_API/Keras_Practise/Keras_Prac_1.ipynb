```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Activation,Dense
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.metrics import categorical_crossentropy


import random
import numpy as np

from sklearn.preprocessing import MinMaxScaler

import warnings
warnings.filterwarnings(action='ignore')
```

```python
train_samples = []
train_labels = []

for i in range(50):
    random_young = random.randint(13,64)
    train_samples.append(random_young)
    train_labels.append(1)
    
    random_old = random.randint(65,100)
    train_samples.append(random_old)
    train_labels.append(0)

for i in range(1000):
    random_young = random.randint(13,64)
    train_samples.append(random_young)
    train_labels.append(0)

    random_old = random.randint(65,100)
    train_samples.append(random_old)
    train_labels.append(1)
```

```python

    
    
print('The number of train samples is ',len(train_samples))
```

```python
random.shuffle(train_samples)
random.shuffle(train_labels)

train_samples = np.array(train_samples)
train_labels = np.array(train_labels)

```

```python
scaler = MinMaxScaler(feature_range=(0,1))
scaled_train_samples = scaler.fit_transform(train_samples.reshape(-1,1))
```

```python
print('View of train samples before scaling is ',train_samples[0:5])
print('View of train samples after scaling is ',scaled_train_samples[0:5])
```
