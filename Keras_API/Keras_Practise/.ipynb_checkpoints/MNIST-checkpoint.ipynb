```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from tensorflow import keras

from tensorflow.math import confusion_matrix
from keras.datasets import mnist
```

```python
(X_train,y_train),(X_test,y_test) = mnist.load_data()
```

```python
print("The dimension of train data ",X_train.shape)
print("The dimension of train data label ",y_train.shape)
print("The dimension of test data ",X_test.shape)
print("The dimension of test data label ",y_test.shape)
```

```python
print('Max pixel value in the image ',np.max(X_train[20]))
print('Min pixel value in the image ',np.min(X_train[20]))
```

```python
X_train = X_train/255
X_test = X_test/255
```

```python
X_train[20]
```

```python
print('Max pixel value in the image ',np.max(X_train[20]))
print('Min pixel value in the image ',np.min(X_train[20]))
```

```python
plt.imshow(X_train[20])
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(units= 50,activation='relu'),
    keras.layers.Dense(units= 50,activation='relu'),
    keras.layers.Dense(units= 10,activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='Adam',
              metrics=['accuracy']
)
```

```python
model.fit(x=X_train,y=y_train,epochs=10,batch_size=10,shuffle=True,validation_split=0.1,verbose=2)
```

```python
model_variant_1 = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(units= 50,activation='tanh'),
    keras.layers.Dense(units= 50,activation='tanh'),
    keras.layers.Dense(units= 50,activation='tanh'),
    keras.layers.Dense(units= 50,activation='tanh'),
    keras.layers.Dense(units= 50,activation='tanh'),
    keras.layers.Dense(units= 10,activation='sigmoid')
])



model_variant_1.compile(loss='sparse_categorical_crossentropy',
              optimizer='Adam',
              metrics=['accuracy']
)

model_variant_1.fit(x=X_train,y=y_train,epochs=10,batch_size=10,shuffle=True,validation_split=0.1,verbose=2)
```
